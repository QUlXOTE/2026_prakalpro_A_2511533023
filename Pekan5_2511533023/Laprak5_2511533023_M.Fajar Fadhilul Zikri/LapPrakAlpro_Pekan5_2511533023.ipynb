### 1. Perulangan (`Looping`)

#### Syntax / Struktur Umum (Python):
```python
for variabel in range(awal, akhir, langkah):
    # Blok kode yang dieksekusi berulang

## 2.2 Kode Program

### 2.2.1 Program "for1_2511533023"


````python
ulang_3023 = int(input("Masukkan jumlah perulangan: "))

for i in range(ulang_3023):
    print(f"Perulangan ke-{i}")
````

**Penjelasan :**
Kode di atas digunakan untuk melakukan perulangan berbasis *counted loop* menggunakan fungsi `input()` dan struktur `for...in range()`. Pertama, program meminta pengguna memasukkan sebuah angka melalui keyboard, lalu konversi `int()` mengubah teks masukan tersebut menjadi tipe data integer dan menyimpannya ke dalam variabel `ulang_3023`. Selanjutnya, fungsi `range(ulang_3023)` menghasilkan urutan nilai numerik mulai dari `0` hingga `ulang_3023 - 1`. Pada setiap iterasi, nilai indeks saat ini disimpan dalam variabel `i` dan dicetak ke layar menggunakan *f-string* dengan format "Perulangan ke-{i}", sehingga teks tersebut akan tampil berulang secara berurutan sebanyak angka yang dimasukkan oleh pengguna.

### 2.2.2 Program  "for2_2511533023"

````python
ulang_3023 = int(input("Masukkan Perulangan: "))
print("Perulangan ke-0 sampai ke-", ulang_3023)
for i in range(ulang_3023):
    print(i, end=" ")
print()
print("Perulangan ke-1 sampai ke-", ulang_3023)
for i in range(1,ulang_3023 + 1):
    print(i, end=" ")
````

**Penjelasan program  di atas:**
Kode tersebut digunakan untuk menampilkan urutan angka perulangan berdasarkan jumlah yang diinputkan pengguna, dengan dua jenis rentang (*range*) indeks yang berbeda. Pertama, program menerima masukan angka dari pengguna dan mengonversinya menjadi integer dalam variabel `ulang_3023`. Pada bagian pertama, program melakukan perulangan dari indeks `0` hingga `ulang_3023 - 1` menggunakan `range(ulang_3023)` dan mencetak nilainya secara mendatar (dipisahkan spasi). Pada bagian kedua, program menampilkan deret angka yang dimulai dari `1` hingga tepat bernilai `ulang_3023` dengan memanfaatkan fungsi `range(1, ulang_3023 + 1)`. Hasil akhirnya adalah dua baris output: baris pertama menunjukkan rentang angka berbasis indeks nol (0-indexed), sedangkan baris kedua menunjukkan rentang angka berbasis penomoran alami mulai dari angka 1.



### 2.2.3 Class "for3_2511533023"

````python
ulang_3023 = int(input("Masukkan jumlah perulangan: "))
jumlah_3023 = 0
for i in range(1, ulang_3023 + 1):
    print(i, end=" ")
    jumlah_3023 = jumlah_3023 + i

    if i < ulang_3023:
        print(" + ", end="")
    else:
        print(" = ", jumlah_3023, end="")
print()
print("jumlah = ", jumlah_3023)
````

**Penjelasan Program:**
Kode tersebut digunakan untuk menghitung sekaligus menampilkan deret penjumlahan angka berurutan dari `1` hingga batas angka yang diinputkan oleh pengguna. Pertama, nilai input dikonversi menjadi tipe data integer dan disimpan dalam variabel `ulang_3023`, serta variabel `jumlah_3023` diinisialisasi dengan nilai `0` untuk menampung total akumulasi. Melalui perulangan `for` dengan `range(1, ulang_3023 + 1)`, program mencetak nilai `i` secara mendatar dan menyimpannya ke dalam akumulator `jumlah_3023`. Di dalam perulangan, percabangan `if-else` digunakan untuk mengatur tampilan visual: karakter tanda tambah (`+`) akan dicetak di antara angka-angka deret, sedangkan tanda sama dengan (`=`) beserta hasil akumulasi sementara akan dicetak ketika perulangan mencapai angka terakhir. Di akhir program, hasil total penjumlahan akhir dicetak kembali di baris baru sebagai konfirmasi total akumulasi seluruh nilai deret.

### 2.2.4 Class "nested_for1_2511533023"

````python
batas_3023 = int(input("Masukkan nilai batas: "))
for line_3023 in range(1, batas_3023 + 1):
    for j in range(1, (-1 * line_3023 + batas_3023) + 1):
        print(".", end="")
    print(line_3023)
````

**Penjelasan Program:**
Kode tersebut digunakan untuk membuat pola cetak angka dengan barisan tanda titik di bagian depannya menggunakan struktur *nested loop* (perulangan bersarang). Program diawali dengan meminta masukan batas dari pengguna yang disimpan sebagai tipe integer pada variabel `batas_3023`. Perulangan luar (*outer loop*) mengontrol baris dari nilai `1` hingga `batas_3023`, sedangkan perulangan dalam (*inner loop*) mencetak tanda titik (`.`) tanpa pindah baris. Jumlah titik yang dicetak pada setiap baris dihitung melalui rumus `(-1 * line_3023 + batas_3023)`, yang menghasilkan pola jumlah titik yang berkurang satu di setiap kenaikan baris. Setelah perulangan dalam selesai, program mencetak angka baris saat ini (`line_3023`) sekaligus berpindah ke baris baru, sehingga membentuk pola rata kanan berupa urutan angka yang diawali oleh deretan titik secara menurun.

### 2.2.5 Class "nested_for2_2511533023"

````python
batas_3023 = int(input("Masukkan nilai batas: "))
for i in range(1, batas_3023 + 1):
    for j in range(1, batas_3023 + 1):
        print("*", end="")
    print()
````

**Penjelasan Program:**
Kode tersebut digunakan untuk membuat pola persegi atau kisi-kisi bintang berukuran $N \times N$ menggunakan struktur *nested loop* (perulangan bersarang). Program menerima nilai batas bertipe integer dari pengguna yang disimpan dalam variabel `batas_3023` untuk menentukan dimensi persegi. Perulangan luar (*outer loop*) dengan variabel `i` bertugas mengatur jumlah baris dari `1` hingga `batas_3023`, sedangkan perulangan dalam (*inner loop*) dengan variabel `j` bertugas mencetak karakter bintang (`*`) secara horizontal tanpa pindah baris sebanyak nilai batas tersebut. Setelah perulangan dalam selesai mencetak satu baris bintang penuh, perintah `print()` dipanggil untuk berpindah ke baris baru, sehingga menghasilkan output berupa blok persegi bintang dengan jumlah baris dan kolom yang sama sesuai masukan pengguna.

### 2.2.6 class "nested_for3_2511533023"

````python
batas_3023 = int(input("Masukkan nilai batas: "))
for i in range(batas_3023 + 1):
    for j in range(batas_3023 + 1):
        print(i + j, end=" ")
    print() 
````

**Penjelasan Program:**
Kode tersebut digunakan untuk membuat tabel penjumlahan atau matriks matriks nilai $(i + j)$ berukuran $(N+1) \times (N+1)$ menggunakan struktur *nested loop* (perulangan bersarang). Program diawali dengan meminta masukan angka dari pengguna yang disimpan dalam variabel `batas_3023`. Perulangan luar (*outer loop*) dengan variabel `i` mengontrol indeks baris mulai dari `0` hingga `batas_3023`, sedangkan perulangan dalam (*inner loop*) dengan variabel `j` mengontrol indeks kolom dalam rentang nilai yang sama. Pada setiap sel, program mencetak hasil penjumlahan indeks baris dan kolom (`i + j`) secara horizontal yang dipisahkan oleh spasi. Setelah seluruh kolom pada satu baris selesai dicetak, perintah `print()` tanpa argumen dijalankan untuk berpindah ke baris berikutnya, sehingga menghasilkan tampilan matriks angka yang nilainya meningkat secara bertahap dari sudut kiri atas ke sudut kanan bawah.

### 2.2.7 Class nested_for4_2511533023

````python
tinggi_3023 = int(input("Masukkan tinggi pola (bilangan genap, misalnya 10): "))

if tinggi_3023 % 2 != 0:
    print("Tinggi harus bilangan genap")
else:
    a_3023 = tinggi_3023 - 2
    c_3023 = a_3023
    lebar_3023 = (2 * tinggi_3023) - 2

    for i_3023 in range(1, tinggi_3023 + 1):
        b_3023 = c_3023 + 1

        for j_3023 in range(1, lebar_3023 + 1):

            # Baris atas dan bawah
            if i_3023 == 1 or i_3023 == tinggi_3023:
                if j_3023 == 1 or j_3023 == lebar_3023:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris isi
            else:
                if j_3023 == 1 or j_3023 == lebar_3023:
                    print("|", end="")
                else:
                    if j_3023 == c_3023:
                        print("<", end="")
                    elif j_3023 == b_3023:
                        print(">", end="")
                    elif j_3023 == (lebar_3023 - c_3023):
                        print("<", end="")
                    elif j_3023 == (lebar_3023 - c_3023 + 1):
                        print(">", end="")
                    elif j_3023 > b_3023 and j_3023 < (lebar_3023 - c_3023):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_3023 -= 2

        if a_3023 <= 0:
            c_3023 = (-a_3023) + 2
        else:
            c_3023 = a_3023
````

**Penjelasan Singkat:**
Kode tersebut digunakan untuk membuat pola bingkai geometris bersimetri horizontal dan vertikal menggunakan percabangan dan perulangan bersarang (*nested loop*). Pertama, program memeriksa apakah tinggi yang diinputkan pengguna (`tinggi_3023`) merupakan bilangan genap; jika ganjil, program menampilkan pesan peringatan. Jika valid, program menghitung lebar bingkai dan menggunakan perulangan luar `i_3023` untuk baris serta perulangan dalam `j_3023` untuk kolom. Baris pertama dan terakhir dicetak sebagai pembatas atas/bawah menggunakan kombinasi karakter `#` di sudut dan `=` di tengah, sedangkan baris-baris tengah dicetak dengan pembatas samping `|`. Di dalam area isi bingkai, posisi karakter khusus seperti `<` dan `>` serta titik `.` atau spasi ditentukan secara dinamis berdasarkan variabel kontrol `c_3023` yang nilainya terus berkurang lalu berbalik naik setelah melewati pertengahan tinggi, sehingga membentuk pola simetris berbentuk intan/jam pasir di dalam bingkai.

### 2.2.8 Class Tugasprak5_2511533023

````python
tinggi_3023 = int(input("Masukkan pola: "))

for i_3023 in range(1, tinggi_3023 + 1):
    for j_3023 in range(tinggi_3023 - i_3023):
        print(" ", end="")
    for k_3023 in range(2 * i_3023 - 1):
        print("*", end="")
    print()
````

**Penjelasan Singkat:**
Kode tersebut digunakan untuk mencetak pola piramida simetris berisi bintang (`*`) dengan tinggi sesuai input yang dimasukkan pengguna. Program diawali dengan meminta nilai integer `tinggi_3023` untuk menentukan jumlah baris piramida. Perulangan luar (`i_3023`) mengontrol urutan baris dari `1` hingga `tinggi_3023`. Pada setiap baris, program menggunakan dua perulangan dalam secara berurutan: perulangan pertama (`j_3023`) mencetak spasi sebanyak `tinggi_3023 - i_3023` untuk menggeser bintang ke tengah, sedangkan perulangan kedua (`k_3023`) mencetak karakter bintang sebanyak `2 * i_3023 - 1` (jumlah ganjil: 1, 3, 5, dan seterusnya). Kombinasi spasi di awal dan deretan bintang bermuatan ganjil ini menghasilkan bentuk segitiga piramida yang rapi dan simetris di layar.

Berbeda dengan pola dasar, program ini menuntut pembentukan deret angka yang bergerak dinamis menurun dan menaik secara palindromik pada setiap baris serta diselaraskan dengan poros kristal di pusat bangun.

**Kode Jawaban :**

````python
n_3023 = int(input("masukkan angka: "))
print("#", end="")
for i_3023 in range(4 * n_3023 + 5):
    print("=", end="")
print("#", end="")
print()

for baris_3023 in range(n_3023, 0, -1):
        print("| ", end="")
        for spsi_3023 in range(2 * (n_3023 - baris_3023)):
            print(" ", end="")
        for angka_3023 in range(baris_3023, 0, -1):
            print(str(angka_3023) + " ", end="")
        print("<*>", end="")
        for angka_3023 in range(1, baris_3023 + 1):
            print(" " + str(angka_3023), end="")
        for spasi_3023 in range(2 * (n_3023 - baris_3023)):
            print(" ", end="")
        print(" |", end="")
        print()

print("|", end="")
for spasi_3023 in range(2 * n_3023 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_3023 in range(2 * n_3023 + 1):
    print(" ", end="")
print("|", end="")
print()

for baris_3023 in range(1, n_3023 + 1):
        print("| ", end="")
        for spsi_3023 in range(2 * (n_3023 - baris_3023)):
            print(" ", end="")
        for angka_3023 in range(baris_3023, 0, -1):
            print(str(angka_3023) + " ", end="")
        print("<*>", end="")
        for angka_3023 in range(1, baris_3023 + 1):
            print(" " + str(angka_3023), end="")
        for spasi_3023 in range(2 * (n_3023 - baris_3023)):
            print(" ", end="")
        print(" |", end="")
        print()

print("#", end="")
for i_3023 in range(4 * n_3023 + 5):
    print("=", end="")
print("#", end="")
print()
````

Selanjutnya, program masuk ke Fase 1 untuk mencetak paruh atas jam pasir, di mana deret angka bergerak mundur dari $N$ ke $1$ dan naik kembali dari $1$ ke $N$ di sekeliling poros pusat <*>, dengan spasi penyeimbang yang makin bertambah ke arah tengah.
