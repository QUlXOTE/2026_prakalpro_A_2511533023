## 2.2 Kode Program

### 2.2.1 Program "numerik1_2511533023"


```python
# create a variable with integer value.
a_3023 = 100
print("The type of variable having value",a_3023,"is",type(a_3023))

 # create a variable with float value.
b_3023 = 10.2345
print("The type of variable having value",b_3023,"is",type(b_3023))

 # create a variable with complex value.
c_3023 = 100+3j
print("The type of variable having value",c_3023,"is",type(c_3023)) 
```

**Penjelasan :**
1. untuk mencetak type data yang ada pada variable a_3023
2. mencetak type data dari nilai yang terkandung dalam variable b_3023
3. mencetak type data dari nilai yang terkandung dalam variable c_3023

### 2.2.3 Class "Kostanta_2511533023"

```python
from typing import Final
PI_3023: Final[float] = 3.14
print("pi: %f" % (PI_3023))
jari_3023 = float(input('Masukkan nilai jari-jari: '))
luas_3023 = PI_3023 * jari_3023 * jari_3023
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3023, luas_3023))
```

Penjelasan Program:
program kostanta ini menggunakan penanda final bertipe float agar nilai PI_3023 bersifat tetap.

### 2.2.4 Class "Input_2511533023"

```python
# Program ini menggunakan fungsi input() untuk menerima input dari pengguna dan menghitung luas persegi panjang berdasarkan panjang dan lebar yang dimasukkan.
# Nilai yang di masukkkan akan dikonversi menjadi tipe data integer 
panjang_3023 = input("Masukkan nilai panjang: ")
Lebar_3023 = input("Masukkan nilai lebar: ")
luas_3023 = int(panjang_3023) * int(Lebar_3023)
print("Luas =", luas_3023)
```

Penjelasan Program:
pada program ini kita hanya membuat method nya luas persegi panjang kalau kita perhatikan nilai pada panjang merupakan string  tapi hasil outputnya integer hal ini karena luas_3023 mengkonversi dari type data string ke int dengan menambahkan int(..).

### 2.2.5 Class "Boolean_2511533023"

```python
# deklarasi variabel boolean
is_lulus_3023 = True
is_cumlaude_3023 = True

# Menggunakan Boolean
nilai_3023 = 85
batas_lulus_3023 = 75

# Menentukan nilai Boolean berdasarkan kondisi
status_kelulusan_3023 = nilai_3023 >= batas_lulus_3023

print("=== Check Kelulusan ===")
print("Nilai:", nilai_3023)
print("Apakah Lulus:", status_kelulusan_3023)
if is_lulus_3023 and is_cumlaude_3023:
    print("Selamat, Anda Lulus dengan predikat Cum laude!")
```

Penjelasan Program:
Pada type data boolean kita hanya memiliki nilai benar dan salah / True and False jika nila_3023 tidak memenuhi standar kelulusan maka bernilai False dimana program akan memberikan nilai False pada apakah lulus.
