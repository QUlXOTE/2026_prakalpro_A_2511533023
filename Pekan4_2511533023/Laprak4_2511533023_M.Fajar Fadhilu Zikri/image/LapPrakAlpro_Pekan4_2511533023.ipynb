## 2.1 Dasar Teori
Berikut adalah tipe dari percabangan pada bahasa python:

**1. Percabangan *If* Tunggal**

Struktur *if* digunakan untuk mengevaluasi satu kondisi tunggal yang akan mengeksekusi blok kode di dalamnya hanya ketika nilai kondisi tersebut bernilai benar (*True*). Jika kondisi bernilai salah (*False*), maka program akan langsung melompati blok instruksi tersebut dan melanjutkan ke baris berikutnya. Penerapan bentuk ini sangat cocok untuk perintah validasi sederhana yang tidak memerlukan penanganan kondisi alternatif.

**2. Percabangan *If-Else***

Struktur *if-else* menyediakan alur dua jalur keputusan untuk menangani situasi di mana program harus memilih salah satu dari dua tindakan yang saling bertentangan. Blok *if* akan dijalankan jika syarat utama terpenuhi, sedangkan blok *else* berfungsi sebagai penampung otomatis apabila kondisi tersebut bernilai salah. Mekanisme ini memastikan bahwa salah satu dari dua blok kode pasti dieksekusi berdasarkan evaluasi kondisi yang diberikan.

**3. Percabangan *If-Elif-Else***

Struktur *if-elif-else* dirancang untuk menangani evaluasi banyak kondisi bertingkat secara berurutan hingga ditemukan satu kondisi pertama yang bernilai benar. Ketika salah satu kondisi *if* atau *elif* terpenuhi, program akan langsung mengeksekusi bloknya dan mengabaikan seluruh pengujian kondisi di bawahnya. Blok *else* opsional diletakkan di akhir sebagai jalur pemungkas jika tidak ada satupun kondisi sebelumnya yang memenuhi syarat.

**4. Percabangan *Multi-IF* Terpisah**

Struktur *multi-if* terdiri dari serangkaian instruksi *if* independen yang dievaluasi secara terpisah satu sama lain dari atas ke bawah tanpa saling menggantikan. Berbeda dengan *elif*, setiap blok *if* berdiri sendiri sehingga beberapa kondisi dapat bernilai benar secara bersamaan dalam satu kali eksekusi. Konsep ini sangat ideal diterapkan pada kasus akumulasi nilai, seperti perhitungan penumpukan diskon atau kriteria kualifikasi ganda.

**5. Percabangan *Match-Case***

Struktur *match-case* merupakan fitur pencocokan pola (*pattern matching*) yang menyederhanakan pemilihan keputusan berdasarkan kesesuaian nilai spesifik dari sebuah variabel. Evaluasi dilakukan dengan membandingkan nilai input terhadap setiap opsi *case* secara lebih efisien dan rapi dibandingkan deretan *elif* yang panjang. Penulisan `case _` disediakan di bagian akhir sebagai penangan default untuk menangkap nilai yang tidak cocok dengan opsi manapun.

## 2.2 Kode Program

### 2.2.1 Program "if1_2511533023"


```python
ipk_3023 = float(input("Input IPK Anda = "))

if ipk_3023 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK" + str(ipk_3023))

print("Program Selesai")
```

**Penjelasan :**
Kode di atas memeriksa kualifikasi kelulusan dengan predikat "Sangat Memuaskan" menggunakan struktur percabangan `if` tunggal. Program mengonversi input `ipk_3023` menjadi bilangan desimal (*float*) lalu mengevaluasi apakah nilainya lebih besar dari `2.75`. Jika syarat tersebut terpenuhi, pesan kelulusan beserta nilai IPK akan dicetak, dan instruksi "Program Selesai" akan selalu dieksekusi di akhir.

### 2.2.2 Program  "if2_2511533023"

```python
ipk_3023 = float(input("Input IPK Anda = "))

if ipk_3023 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK" + str(ipk_3023))
else:
    print("Anda tidak lulus")
print("Program Selesai")
```

**Penjelasan program  di atas:**
Kode di atas menentukan status kelulusan akademik mahasiswa berdasarkan nilai IPK menggunakan struktur percabangan `if-else`. Program mengonversi input `ipk_3023` menjadi bilangan desimal (*float*) untuk dibandingkan dengan ambang batas `2.75`. Jika IPK lebih dari 2.75, program mencetak pesan lulus "Sangat Memuaskan", sedangkan jika tidak, pesan "Anda tidak lulus" akan ditampilkan sebelum program berakhir.



### 2.2.3 Class "if_else1_2511533023"

```python
umur_3023 = int(input("Input umur Anda: "))
sim_3023 = input("Apakah anda sudah punya Sim c: ")[0]

if umur_3023 >= 17 and sim_3023 == 'y':
    print("Anda sudah dewasa dan boleh bawa motor")
elif umur_3023 >= 17 and sim_3023 != 'y':
    print("Anda sudah dewasa tetapi belum tidak boleh bawa motor")
elif umur_3023 < 17 and sim_3023 == 'y':
    print("Anda belum cukup umur punya sim")
else:
    print("Anda belum cukup umur dan tidak boleh bawa motor")
print("Program Selesai")
```

**Penjelasan Program:**
ode di atas digunakan untuk memverifikasi izin mengendarai sepeda motor berdasarkan umur dan kepemilikan SIM C menggunakan empat blok `if` independen. Program akan mengevaluasi kombinasi variabel `umur_3023` dan `sim_3023` secara logis menggunakan operator `and` serta operator perbandingan (`>=`, `<`, `==`, `!=`). Teks peringatan atau izin yang sesuai hanya akan dicetak pada konsol ketika kondisi kriteria spesifik pengguna bernilai benar (*True*).

### 2.2.4 Class "multi_if1_2511533023"

```python
umur_3023 = int(input("Input umur Anda: "))
sim_3023 = input("Apakah anda sudah punya Sim c (y/t): ")[0]

if umur_3023 >= 17 and sim_3023 == 'y':
    print("Anda sudah dewasa dan boleh bawa motor")
if umur_3023 >= 17 and sim_3023 != 'y':
    print("Anda sudah dewasa tetapi belum tidak boleh bawa motor")
if umur_3023 < 17 and sim_3023 == 'y':
    print("Anda belum cukup umur punya sim")
if umur_3023 < 17 and sim_3023 != 'y':
    print("Anda belum cukup umur bawa motor")
```

**Penjelasan Program:**
Kode di atas menguji empat kondisi kelayakan membawa motor menggunakan serangkaian instruksi `if` independen dan operator logika `and`. Program mengevaluasi kombinasi nilai `umur_3023` dan status `sim_3023` secara eksplisit untuk menentukan satu pesan status yang sesuai. Hanya blok yang syarat logikanya bernilai *True* yang akan dieksekusi untuk mencetak output ke layar.

### 2.2.5 Class "multi_if2_2511533023"

```python
total_belanja_3023 = float(input("Masukkan total belanja (Rp): "))

input_member_3023 = input("Apakah anda member? (y/t): ").strip().lower()
is_member_3023 = input_member_3023 in ["y", "ya"]

input_promo_3023 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_3023 = input_promo_3023 in ["y", "ya"]

total_diskon_persen_3023 = 0

if total_belanja_3023 > 1000000:
    total_diskon_persen_3023 += 10
if is_member_3023:
    total_diskon_persen_3023 += 5
if kode_promo_valid_3023:
    total_diskon_persen_3023 += 15

nominal_diskon_3023 = total_belanja_3023 * (total_diskon_persen_3023 / 100)
total_bayar_3023 = total_belanja_3023 - nominal_diskon_3023

print("\n---Rincian Pembayaran---")
print(f"Total Diskon : {total_diskon_persen_3023}% (Rp {nominal_diskon_3023:,.0f})")
print(f"Total Bayar : Rp {total_bayar_3023:,.0f}")

print(f"Total Diskon yang anda dapatkan: {total_diskon_persen_3023}%")
```

**Penjelasan Program:**
Kode di atas menghitung total pembayaran dan akumulasi diskon belanja dengan menerapkan tiga blok `if` terpisah secara independen. Program mengevaluasi kriteria total belanja, status keanggotaan, dan kevalidan kode promo untuk menambah persentase diskon secara bertingkat. Hasil akhir berupa persentase diskon, potongan harga, dan total bayar kemudian ditampilkan secara rapi menggunakan *f-string*.

Blok kode akan mencetak huruf mutu ('A' hingga 'D') pada kondisi pertama yang bernilai *True*, atau mencetak 'E' melalui blok `else` jika nilai di bawah 50.

# 2.2.7 Class match_case1_2511533023

```python
bulan_3023 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_3023:
    case 1:
        print("Januari")
    case 2: 
        print("Februari")
    case 3:
        print("maret")
    case 4: 
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasan Singkat:**
Kode di atas mengonversi masukan angka bulan menjadi nama bulan yang sesuai menggunakan struktur percabangan `match-case`. Program akan mencocokkan nilai dari variabel `bulan_3023` (dari angka 1 hingga 12) dengan setiap pilihan *case* secara spesifik. Jika angka yang dimasukkan berada di luar rentang 1 hingga 12, maka blok `case _` akan dieksekusi untuk menampilkan pesan "Angka tidak valid".

## Tugas Praktikum

**Soal :** 
 Sebuah wahana rekreasi petualangan membutuhkan sistem loket berbasis konsol untuk menentukan jenis wahana, memverifikasi kelaikan pengunjung dengan kondisi majemuk (logical operators), menghitung diskon akumulatif menggunakan multi-if terpisah, serta mencetak audit status operasional.

**Kode Jawaban :**

a. Kode Program "tugas3_2511533023"

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# 1. Input Data Pengunjung & String Handling
nama_3023 = input("Masukkan Nama Pengunjung        : ").strip()
umur_3023 = int(input("Input umur anda                  : "))

sim_input_3023 = input("Apakah Anda Sudah Punya SIM C (y/t): ").strip().lower()
sim_3023 = sim_input_3023[0] if sim_input_3023 else 't'

# Menu Paket Wahana
print("\nPilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")

pilihan_paket_3023 = int(input("Masukkan nomor paket (1-5)      : "))
jumlah_tiket_3023 = int(input("Masukkan jumlah tiket            : "))

# Penerapan IF Tunggal untuk Validasi Kelogisan Tiket
if jumlah_tiket_3023 <= 0:
    print("\n[Peringatan] Kuota tiket tidak valid! Jumlah tiket harus lebih dari 0.")

# 2. Pemilihan Wahana Menggunakan match - case
nama_wahana_3023 = ""
harga_satuan_3023 = 0
valid_paket_3023 = True

match pilihan_paket_3023:
    case 1:
        nama_wahana_3023 = "Wahana Safari Rimba"
        harga_satuan_3023 = 50000
    case 2:
        nama_wahana_3023 = "Wahana Arung Jeram"
        harga_satuan_3023 = 75000
    case 3:
        nama_wahana_3023 = "Wahana Motor ATV Ekstrim"
        harga_satuan_3023 = 120000
    case 4:
        nama_wahana_3023 = "Wahana Roller Coaster Kilat"
        harga_satuan_3023 = 100000
    case 5:
        nama_wahana_3023 = "Wahana All-Access VIP"
        harga_satuan_3023 = 220000
    case _:
        print("\nPaket wahana tidak valid!")
        valid_paket_3023 = False

# Jalankan sisa transaksi hanya jika pilihan paket valid dan tiket valid
if valid_paket_3023 and jumlah_tiket_3023 > 0:
    is_member_input_3023 = input("Apakah Anda member? (y/t)        : ").strip().lower()
    kode_promo_input_3023 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()

    # 3. Validasi Izin Kendali Wahana Menggunakan if - elif - else dan Operator Logika
    print("\n--- KELAYAKAN PENGENDARA WAHANA ---")
    if pilihan_paket_3023 == 3:
        if umur_3023 >= 17 and sim_3023 == 'y':
            print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
        elif umur_3023 >= 17 and sim_3023 != 'y':
            print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
        elif umur_3023 < 17 and sim_3023 == 'y':
            print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
        else:
            print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
    else:
        if umur_3023 >= 10:
            print(f"Status Akses: Anda memenuhi syarat umur untuk menikmati {nama_wahana_3023}.")
        else:
            print(f"Status Akses: Anda belum cukup umur untuk menikmati {nama_wahana_3023}.")

    # 4. Akumulasi Diskon Bertingkat Menggunakan Multi-IF Terpisah
    subtotal_3023 = harga_satuan_3023 * jumlah_tiket_3023
    total_diskon_persen_3023 = 0

    if subtotal_3023 >= 200000:
        total_diskon_persen_3023 += 10

    if is_member_input_3023 in ['y', 'ya']:
        total_diskon_persen_3023 += 5

    if kode_promo_input_3023 in ['y', 'ya']:
        total_diskon_persen_3023 += 15

    if jumlah_tiket_3023 >= 5:
        total_diskon_persen_3023 += 5

    # 5. Evaluasi Kelulusan Audit Menggunakan if - else & Output Rincian
    nominal_diskon_3023 = subtotal_3023 * (total_diskon_persen_3023 / 100)
    total_bayar_3023 = subtotal_3023 - nominal_diskon_3023

    print("\n--- Rincian Pembayaran ---")
    print(f"Subtotal Belanja : Rp {subtotal_3023:,.0f}")
    print(f"Total Diskon     : {total_diskon_persen_3023}% (Rp {nominal_diskon_3023:,.0f})")
    print(f"Total Bayar      : Rp {total_bayar_3023:,.0f}")

    if total_bayar_3023 > 300000:
        print("Catatan Layanan  : Selamat! Anda berhak mendapatkan Souvenir Gratis.")
    else:
        print("Catatan Layanan  : Terima kasih telah berkunjung.")

print("\nProgram Selesai")
```

**Penjelasan :**
Program di atas merupakan sistem loket berbasis konsol yang mengintegrasikan berbagai percabangan Python untuk memproses transaksi tempat rekreasi. Program menerima input identitas pengunjung, memvalidasi kelayakan akses wahana berdasarkan umur dan SIM, serta mengolah paket pilihan menggunakan struktur `match-case`. Selanjutnya, perhitungan diskon akumulatif dilakukan secara independen sebelum mencetak rincian biaya akhir beserta bonus souvenir secara otomatis.
